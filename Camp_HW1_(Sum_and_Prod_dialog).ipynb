```python
def generate_pairs():
    """
    Generate all possible pairs (a, b) for 1 < a < b < 100.
    --------
    Returns:
    list of tuple: where each tuple represents a valid pair (a, b)
    """
    pairs = []
    for a in range(2, 100):
        for b in range(a + 1, 100):
            pairs.append((a, b))
    return pairs

pairs = generate_pairs()
print(f"count of all posible pairs: {len(pairs)}")
```

```python
def build_mappings(pairs):
    """
    build two dictionaries:
    - prod_to_pairs: for each product, we store a list of pairs that produce it;
    - sum_to_pairs: for each sum, we store the list of pairs that produce it.
    -----
    Args:
        pairs (list of tuple): A list of integer pairs (a, b).
    -----
    Returns:
        tuple: A tuple containing two dictionaries: (prod_to_pairs, sum_to_pairs).
    """
    prod_to_pairs = {}
    sum_to_pairs = {}
    for a, b in pairs:
        prod = a * b
        s = a + b
        prod_to_pairs.setdefault(prod, []).append((a, b))
        sum_to_pairs.setdefault(s, []).append((a, b))
    return prod_to_pairs, sum_to_pairs

prod_to_pairs, sum_to_pairs = build_mappings(pairs)

print(f"count of all posible product pairs: {len(prod_to_pairs)}")
print(f"count of all posible sum pairs: {len(sum_to_pairs)}")
```

```python
def step1_filter(pairs, prod_to_pairs):
    """
    Step 1. Prod: i don`t know the numbers.
    reject those pairs in which the product can be obtained in only one way
    -----
    Args:
        pairs (list of tuple): A list of candidate pairs (a, b).
        prod_to_pairs (dict): A dictionary mapping each product to the list of pairs that yield it
    -----
    Returns:
        list of tuple: where each pair's product is ambiguous
    """
    return [pair for pair in pairs if len(prod_to_pairs[pair[0] * pair[1]]) > 1]

candidates = step1_filter(pairs, prod_to_pairs)
print(f"count of all posible candidates after first step: {len(candidates)}")
```

```python
def step2_filter(pairs, sum_to_pairs, prod_to_pairs):
    """
    Step 2. Sum: I know it.
    check that the product doesn`t have a unique decomposition for each pair of the given sum
    -----
    Args:
        pairs (list of tuple): A list of candidate pairs (a, b) from the previous step
        sum_to_pairs (dict): A dictionary mapping each sum to the list of pairs that yield it
        prod_to_pairs (dict): A dictionary mapping each product to the list of pairs that yield it
    -----
    Returns:
        list of tuple: A filtered list of pairs meeting the condition that every pair with the same
        sum results in an ambiguous product
    """
    filtered = []
    for pair in pairs:
        s = pair[0] + pair[1]
        valid = True
        for cand in sum_to_pairs[s]:
            p = cand[0] * cand[1]
            if len(prod_to_pairs[p]) == 1:  # There is a unique way to get the product
                valid = False
                break
        if valid:
            filtered.append(pair)
    return filtered

candidates = step2_filter(candidates, sum_to_pairs, prod_to_pairs)
print(f"count of all posible candidates after second step: {len(candidates)}")
```

```python
def step3_filter(pairs):
    """
    Step 3. Prod: then i know the numbers.
    there must be only one option left for the product from the received candidates,
    otherwise Prod would not be able to determine the numbers
    -----
    Args:
        pairs (list of tuple): A list of candidate pairs (a, b) after previous filtering steps
    -----
    Returns:
        list of tuple: A list of pairs for which the product uniquely identifies the numbers
    """
    filtered = []
    for pair in pairs:
        p = pair[0] * pair[1]
        # Among the candidates from step 2, we look for those that give the same product
        same_prod = [cand for cand in pairs if cand[0] * cand[1] == p]
        if len(same_prod) == 1:
            filtered.append(pair)
    return filtered

candidates = step3_filter(candidates)
print(f"count of all posible candidates after third step: {len(candidates)}")
```

```python
def step4_filter(pairs, sum_to_pairs):
    """
    Step 4. Sum: then i know the numbers too.
    filters the candidate pairs by ensuring that for each pair,
    its sum is unique among the candidates
    -----
    Args:
        pairs (list of tuple): A list of candidate pairs (a, b) after previous filtering steps
        sum_to_pairs (dict): A dictionary mapping each sum to the list of pairs that yield it

    Returns:
        list of tuple: A list of pairs for which the sum uniquely determines the numbers
    """
    filtered = []
    for pair in pairs:
        s = pair[0] + pair[1]
        same_sum = [cand for cand in pairs if cand[0] + cand[1] == s]
        if len(same_sum) == 1:
            filtered.append(pair)
    return filtered


candidates = step4_filter(candidates, sum_to_pairs)
for candidate in candidates:
  print(f"a = {candidate[0]}")
  print(f"b = {candidate[1]}")
```
